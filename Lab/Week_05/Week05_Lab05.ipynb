# Implementation of DFS   **🔎**

image.png

DFS traverse could start from 0 to 2 to 1 to 3 to 4 to 2, but 2 is visited and 4 also doesn’t have any adjacent nodes that are not visited, so it starts backtracking from 4 to 3 to 1 to 2 to 0.

image.png

```python
def dfs(node, graph, visited, component):
  component.append(node)
  visited[node] = True

  for neighbor in graph[node]:
    if not visited[neighbor]:
      dfs(neighbor, graph, visited, component) # Recursive Call

if __name__ == '__main__':
# Start of main function...
  graph = {
      0: [2],
      1: [2, 3],
      2: [0, 1, 4],
      3: [1, 4],
      4: [2, 3]
  }
  node = 0
  visited = [False] * len(graph) # List of all the nodes  marks as false not visited
  components = []
  dfs(node, graph, visited, components)
  print("Following is the Depth-First Search: ", components)
```

## 1.	Write program to perform Depth First Search. This code should print nodes traversal order and adjacent values for every nodes.

image.png

```python
def dfs(visited, graph, node):
    if node not in visited:
        visited.add(node)
        print(f"Visited Node: {node}")
        print(f"   Adjacent values: {graph[node]}")
        # Recursively visit all unvisited neighbors
        for neighbor in graph[node]:
            dfs(visited, graph, neighbor)

if __name__ == "__main__":
    graph = {
        0: [1, 2, 3],
        1: [0],
        2: [0, 3, 4],
        3: [0, 2],
        4: [2]
    }
    # Create a set to keep track of visited nodes
    visited = set()

    print("--- DFS Traversal Order ---")
    dfs(visited, graph, 0)
```

## 2.	Implement Breadth-first search (BFS)

```python
def bfs(graph, start_node):
    queue = [start_node]
    visited = [start_node]
    while queue:
        current_node = queue.pop(0)

        print(f"Visited Node: {current_node}")
        print(f"   Adjacent values: {graph[current_node]}")

        for neighbor in graph[current_node]:
            if neighbor not in visited:
                visited.append(neighbor)
                queue.append(neighbor)

if __name__ == "__main__":
        graph = {
        0: [1, 2, 3],
        1: [0],
        2: [0, 3, 4],
        3: [0, 2],
        4: [2]
    }
print("--- BFS Traversal Order ---")
bfs(graph, 2)
```
