```python
# Print First 20 Prime Numbers from a Specific Starting Point Defined by user.

def is_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True


start_num = int(input("Enter the starting number to find primes from: "))

print(f"\nFinding the first 20 prime numbers starting from {start_num}:")

count = 0
primes_found = []

while count < 20:
    if is_prime(start_num):
        primes_found.append(start_num)
        count += 1
    start_num += 1

print(primes_found)
```
