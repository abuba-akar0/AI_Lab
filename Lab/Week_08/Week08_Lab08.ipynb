```python
import pandas as pd
import numpy as np
import io
```

## DATASET 🉑

```python
csv_data = """StudentID,Name,Gender,Age,Department,GPA,Attendance,Quiz1,Quiz2,FinalExam
1,Ayesha,Female,20,CS,3.7,85,8,9,88
2,Ali,Male,22,SE,2.9,72,5,6,70
3,Sara,Female,21,CS,3.2,95,7,8,83
4,Usman,Male,23,IT,2.5,65,6,5,68
5,Anum,Female,20,CS,3.9,90,9,10,92
6,Ahmed,Male,24,SE,2.7,60,4,5,63
7,Hira,Female,22,IT,3.4,80,7,8,79
8,Kamran,Male,23,CS,3.0,75,6,6,72
9,Mahnoor,Female,21,SE,3.5,88,8,9,87
10,Zain,Male,22,CS,2.8,55,5,4,60"""

df = pd.read_csv(io.StringIO(csv_data))
df
```

## TASK 1: DATA EXPLORATION 🀄

```python
print("=" * 55)
print("TASK 1: DATA EXPLORATION")
print("=" * 55)

print("\n[1.1] Full DataFrame:")
print(df)

print("\n[1.2] Summary Statistics - describe():")
print(df.describe())

print("\n[1.3] Structure Info - info():")
df.info()
```

## TASK 2: DATA CLEANING 🍽

```python
print("\n" + "=" * 55)
print("TASK 2: DATA CLEANING")
print("=" * 55)

# 2a. Check & remove duplicates
print(f"\n[2.1] Duplicate rows found: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print("      drop_duplicates() applied — dataset is clean.")

# 2b. Check & handle missing values
print(f"\n[2.2] Missing values per column:")
print(df.isnull().sum())
# Fill missing numeric values with column mean (good practice)
df.fillna(df.mean(numeric_only=True), inplace=True)
print("      fillna(mean) applied to handle any future NaN values.")

# 2c. Rename FinalExam to Final_Exam_Score
df.rename(columns={'FinalExam': 'Final_Exam_Score'}, inplace=True)
print("\n[2.3] Renamed 'FinalExam' → 'Final_Exam_Score'")
print("      Updated columns:", list(df.columns))
```

## TASK 3: FEATURE ENGINEERING 🖼

```python
print("\n" + "=" * 55)
print("TASK 3: FEATURE ENGINEERING")
print("=" * 55)

# 3a. Performance_Level based on GPA
def classify_performance(gpa):
    if gpa >= 3.5:
        return "Excellent"
    elif gpa >= 3.0:
        return "Good"
    else:
        return "Needs Improvement"

df['Performance_Level'] = df['GPA'].apply(classify_performance)

# 3b. Total_Quiz = Quiz1 + Quiz2
df['Total_Quiz'] = df['Quiz1'] + df['Quiz2']

print("\n[3.1] New Columns Added — Performance_Level & Total_Quiz:")
print(df[['Name', 'GPA', 'Performance_Level', 'Quiz1', 'Quiz2', 'Total_Quiz']].to_string(index=False))
```

## TASK 4: SORTING & GROUPING 💭

```python
print("\n" + "=" * 55)
print("TASK 4: SORTING & GROUPING")
print("=" * 55)

# 4a. Sort by GPA descending, then Attendance ascending
sorted_df = df.sort_values(by=['GPA', 'Attendance'], ascending=[False, True])
print("\n[4.1] Sorted by GPA (Descending) → Attendance (Ascending):")
print(sorted_df[['Name', 'Department', 'GPA', 'Attendance', 'Performance_Level']].to_string(index=False))

# 4b. Group by Department — avg GPA, avg Attendance, count
grouped = df.groupby('Department').agg(
    Average_GPA=('GPA', 'mean'),
    Average_Attendance=('Attendance', 'mean'),
    Student_Count=('Name', 'count')
).round(2)

print("\n[4.2] Grouped by Department:")
print(grouped)
```

## BONUS: Students above dept average Final Exam 🧪

```python
print("\n" + "=" * 55)
print("BONUS CHALLENGE")
print("=" * 55)

# Use transform to get each student's dept average inline
dept_avg_score = df.groupby('Department')['Final_Exam_Score'].transform('mean')
above_avg = df[df['Final_Exam_Score'] > dept_avg_score]

print("\n[Bonus] Students with Final_Exam_Score ABOVE their Department Average:")
print(above_avg[['Name', 'Department', 'Final_Exam_Score']].to_string(index=False))

print("\n[Dept Averages for reference]:")
print(df.groupby('Department')['Final_Exam_Score'].mean().round(2))
```
