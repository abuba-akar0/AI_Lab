# Python List 🐍

### In this Lab we will be try following list methods with examples ❗

```python
print("Car Models \n")
car_models = [
    "Toyota Camry",
    "Honda Civic",
    "Ford Mustang",
    "Chevrolet Corvette",
    "BMW X5",
    "Audi A4",
    "Mercedes-Benz C-Class",
    "Jeep Wrangler",
    "Subaru Outback",
    "Tesla Model 3"
]

for models in car_models:
  print(models)
```

### append() ⚓

> Adds a single element to the end of the list.



```python
car_models.append("Hyundai Sonata    new")

for models in car_models:
  print(models)
```

### clear()

> Removes all elements, leaving the list empty but keeping the variable reference



```python
car_models.clear()

print(car_models)
```

### copy()

> Returns a shallow copy of the list so changes to one don't affect the other.



```python

new_copy = car_models.copy()
print(new_copy)
```

### count()

> Returns how many times a specific value appears in the list.



```python
print(car_models.count("Audi A4"))
```

### extend()


> Adds all elements of an iterable (like another list) to the end of the current list.





```python
lux_cars = ["Mercedes", "Lexus"]
car_models.extend(lux_cars)
print(car_models)  # Output: ['Mercedes', 'Lexus']
```

### index()

> Returns the index of the first element with the specified value.



```python
print(car_models.index("Audi A4"))  # Output: 1
```

### insert()

> Adds an element at a specific position (index).



```python
car_models.insert(1, "Porsche")
print(car_models)
```

### pop()

> Removes and returns the element at a specific position (defaults to the last item).



```python
removed_car = car_models.pop(2)
print(removed_car)  # Output: Rivian
```

### remove()

> Removes the first occurrence of a specific value.



```python
car_models.remove("BMW X5")
print(car_models)
```

### reverse()

> Reverses the order of the elements in the list in-place.



```python
car_models.reverse()
print(car_models)
```

### sort()

> Sorts the list alphabetically or numerically by default.


```python
car_models.sort()
print(car_models)
```

## Bonus Task: GPA Calculator 🔄

```python
print("Simple GPA calculator")

total_courses = int(input("Enter your total courses: "))

total_points = 0
total_credits = 0

for i in range(total_courses):
    print("\nCourse", i + 1)

    course = input("Enter your course name: ")
    credit_hours = int(input("Enter your credit hours: "))
    obtain_marks = int(input("Enter your obtained marks: "))

    if obtain_marks >= 85:
        grade_points = 4.0
    elif obtain_marks >= 80:
        grade_points = 3.7
    elif obtain_marks >= 75:
        grade_points = 3.3
    elif obtain_marks >= 70:
        grade_points = 3.0
    elif obtain_marks >= 65:
        grade_points = 2.7
    elif obtain_marks >= 60:
        grade_points = 2.3
    elif obtain_marks >= 55:
        grade_points = 2.0
    elif obtain_marks >= 50:
        grade_points = 1.7
    else:
        grade_points = 0.0

    total_points += grade_points * credit_hours
    total_credits += credit_hours

gpa = total_points / total_credits
print("\nYour GPA is:", round(gpa, 2))
```
